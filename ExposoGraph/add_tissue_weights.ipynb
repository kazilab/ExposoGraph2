```python
import json
```

```python
graph_data = "map/graph_data.json"
tissue_data = "data/tissue_expression_data.json"
```

```python
with open(graph_data, "r", encoding="utf-8") as file:
    graph_dict = json.load(file)
```

```python
enzyme_no_tissue = []
for i in graph_dict['nodes']:
    if i['type']=='Enzyme' and "tissue_weights" not in i:
        enzyme_no_tissue.append(i['id'])
```

```python
enzyme_no_tissue
```

```python
with open(tissue_data, "r", encoding="utf-8") as file:
    tissue_dict = json.load(file)
```

```python
enzymes_with_weights = tissue_dict['weights'].keys()
```

```python
enzymes_to_add = list(set(enzymes_with_weights) & set(enzyme_no_tissue))
```

```python
enzymes_to_fix = {k:tissue_dict['weights'][k] for k in enzymes_to_add}
```

```python
enzymes_to_fix = {
    enzyme: {tissue: weight / max(weights.values()) for tissue, weight in weights.items()}
    for enzyme, weights in enzymes_to_fix.items()
}
```

```python
enzymes_to_fix
```

```python
graph_dict
```

```python
for node in graph_dict.get("nodes", []):
    enzyme = node.get("id")
    if enzyme in enzymes_to_fix:
        node["tissue_weights"] = enzymes_to_fix[enzyme]
```
